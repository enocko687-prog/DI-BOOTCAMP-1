# Number Guessing Game

Guess the secret number between 1 and 100. You have 7 attempts.

```python
import random

def number_guessing_game():
    random_number = random.randint(1, 100)
    max_attempts = 7
    guessed_correctly = False

    print("I picked a number between 1 and 100.")

    for attempt in range(1, max_attempts + 1):
        guess = int(input(f"Attempt {attempt}/{max_attempts}. Enter your guess: "))

        if guess < random_number:
            print("Too low!")
        elif guess > random_number:
            print("Too high!")
        else:
            print("Congratulations! You guessed the number!")
            guessed_correctly = True
            break

    if not guessed_correctly:
        print(f"Game over! The correct number was {random_number}.")

number_guessing_game()
```
